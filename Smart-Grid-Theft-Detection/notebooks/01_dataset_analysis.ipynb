```python
# ============================================
# CREATE CLEAN DATASET
# ============================================

MIN_VALID_READINGS = 300

# Keep consumers with enough usable history
enough_history = valid_readings >= MIN_VALID_READINGS

# Create cleaned dataframe
clean_df = df.loc[
    enough_history,
    clean_consumption_cols + ["CONS_NO", "FLAG"]
].copy()

# Reset row index
clean_df.reset_index(drop=True, inplace=True)

print("Clean dataset shape:", clean_df.shape)

print("\nConsumers:", clean_df.shape[0])
print("Consumption dates:", len(clean_consumption_cols))

print("\nFLAG distribution:")
print(clean_df["FLAG"].value_counts())

print("\nMissing values:")
print(clean_df[clean_consumption_cols].isna().sum().sum())
```

```python
# ============================================
# PREPARE CHRONOLOGICAL CONSUMPTION DATA
# ============================================

# Convert column names to actual dates
date_index = pd.to_datetime(
    clean_consumption_cols,
    format="%m/%d/%Y"
)

# Make sure dates are chronological
sorted_pairs = sorted(
    zip(date_index, clean_consumption_cols)
)

sorted_dates = [x[0] for x in sorted_pairs]
sorted_consumption_cols = [x[1] for x in sorted_pairs]

print("First date:", sorted_dates[0])
print("Last date:", sorted_dates[-1])
print("Number of dates:", len(sorted_consumption_cols))
```

```python
# Last 30 days = target period
target_cols = sorted_consumption_cols[-30:]

print("Target period:")
print(target_cols[0], "to", target_cols[-1])

# Number of valid readings in the target period for each consumer
target_valid = clean_df[target_cols].notna().sum(axis=1)

print("\nTarget-period valid readings:")
print(target_valid.describe())

print(
    "\nConsumers with at least 15 valid readings:",
    (target_valid >= 15).sum()
)

print(
    "Consumers with fewer than 15 valid readings:",
    (target_valid < 15).sum()
)
```

```python
# ============================================
# FEATURE ENGINEERING WINDOWS
# ============================================

# Final 30 days = prediction target
target_cols = sorted_consumption_cols[-30:]

# 30 days immediately before the target
recent_30_cols = sorted_consumption_cols[-60:-30]

# 7 days immediately before the target
recent_7_cols = sorted_consumption_cols[-37:-30]

# 90 days before the recent 30-day window
previous_90_cols = sorted_consumption_cols[-150:-60]

print("Recent 7 days:",
      recent_7_cols[0], "to", recent_7_cols[-1])

print("Recent 30 days:",
      recent_30_cols[0], "to", recent_30_cols[-1])

print("Previous 90 days:",
      previous_90_cols[0], "to", previous_90_cols[-1])

print("Target 30 days:",
      target_cols[0], "to", target_cols[-1])
```

```python
# ============================================
# CREATE MODEL FEATURES
# ============================================

feature_df = pd.DataFrame(index=clean_df.index)

# Recent behavior
feature_df["mean_7"] = clean_df[recent_7_cols].mean(axis=1)
feature_df["mean_30"] = clean_df[recent_30_cols].mean(axis=1)

# Variability
feature_df["std_30"] = clean_df[recent_30_cols].std(axis=1)

# Longer-term behavior
feature_df["mean_90"] = clean_df[previous_90_cols].mean(axis=1)
feature_df["std_90"] = clean_df[previous_90_cols].std(axis=1)

# Range
feature_df["min_30"] = clean_df[recent_30_cols].min(axis=1)
feature_df["max_30"] = clean_df[recent_30_cols].max(axis=1)

# Missingness itself can be informative
feature_df["missing_ratio_30"] = (
    clean_df[recent_30_cols].isna().mean(axis=1)
)

feature_df["missing_ratio_90"] = (
    clean_df[previous_90_cols].isna().mean(axis=1)
)

# Change in recent behavior
feature_df["recent_change"] = (
    feature_df["mean_7"] - feature_df["mean_30"]
)

# ============================================
# TARGET
# ============================================

feature_df["target_mean_30"] = (
    clean_df[target_cols].mean(axis=1)
)

# Add labels / IDs for evaluation only
feature_df["CONS_NO"] = clean_df["CONS_NO"]
feature_df["FLAG"] = clean_df["FLAG"]

print("Feature dataset shape:", feature_df.shape)
print("\nFirst rows:")
print(feature_df.head())
```

```python
# ============================================
# BAYESIAN LINEAR REGRESSION DATA
# ============================================

import numpy as np

feature_cols = [
    "mean_7",
    "mean_30",
    "std_30",
    "mean_90",
    "std_90",
    "min_30",
    "max_30",
    "missing_ratio_30",
    "missing_ratio_90",
    "recent_change"
]

X = model_df[feature_cols].copy()

# Log-transform the target because of strong right skew
y = np.log1p(model_df["target_mean_30"])

flags = model_df["FLAG"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTransformed target statistics:")
print(y.describe())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test, flag_train, flag_test = train_test_split(
    X,
    y,
    flags,
    test_size=0.20,
    random_state=42,
    stratify=flags
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import BayesianRidge

bayesian_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", BayesianRidge())
])

bayesian_model.fit(X_train, y_train)

print("Bayesian Linear Regression trained successfully!")
```

```python
# ============================================
# BAYESIAN PREDICTION
# ============================================

bayesian_regressor = bayesian_model.named_steps["model"]

X_test_imputed = bayesian_model.named_steps["imputer"].transform(X_test)
X_test_scaled = bayesian_model.named_steps["scaler"].transform(
    X_test_imputed
)

y_pred_log, y_std_log = bayesian_regressor.predict(
    X_test_scaled,
    return_std=True
)

# Convert prediction back to original consumption scale
y_pred = np.expm1(y_pred_log)

# Actual target on original scale
y_actual = np.expm1(y_test)

print("First 10 actual values:")
print(y_actual[:10])

print("\nFirst 10 predicted values:")
print(y_pred[:10])

print("\nFirst 10 uncertainty values (log scale):")
print(y_std_log[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_actual, y_pred)
rmse = np.sqrt(mean_squared_error(y_actual, y_pred))
r2 = r2_score(y_actual, y_pred)

print("\nBayesian Linear Regression")
print("--------------------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
print("Mean uncertainty:", y_std_log.mean())
```

```python
# ============================================
# ENHANCED BEHAVIORAL FEATURES
# ============================================

enhanced_df = model_df.copy()

# Avoid division by zero
EPS = 1e-6

enhanced_df["ratio_7_30"] = (
    enhanced_df["mean_7"] /
    (enhanced_df["mean_30"] + EPS)
)

enhanced_df["ratio_30_90"] = (
    enhanced_df["mean_30"] /
    (enhanced_df["mean_90"] + EPS)
)

enhanced_df["change_30_90"] = (
    enhanced_df["mean_30"] -
    enhanced_df["mean_90"]
)

enhanced_df["relative_change_30_90"] = (
    (enhanced_df["mean_30"] - enhanced_df["mean_90"]) /
    (enhanced_df["mean_90"] + EPS)
)

enhanced_df["cv_30"] = (
    enhanced_df["std_30"] /
    (enhanced_df["mean_30"] + EPS)
)

enhanced_df["max_to_mean_30"] = (
    enhanced_df["max_30"] /
    (enhanced_df["mean_30"] + EPS)
)

print("Enhanced features created.")
print(enhanced_df.shape)

print("\nNew features:")
print([
    "ratio_7_30",
    "ratio_30_90",
    "change_30_90",
    "relative_change_30_90",
    "cv_30",
    "max_to_mean_30"
])
```

```python
# ============================================
# SAME PERIOD LAST YEAR
# ============================================

last_year_cols = [
    col for col in sorted_consumption_cols
    if pd.to_datetime(col, format="%m/%d/%Y") >= pd.Timestamp("2015-10-02")
    and pd.to_datetime(col, format="%m/%d/%Y") <= pd.Timestamp("2015-10-31")
]

print("Last-year comparison period:")
print(last_year_cols[0], "to", last_year_cols[-1])

enhanced_df["mean_same_period_last_year"] = (
    clean_df.loc[enhanced_df.index, last_year_cols].mean(axis=1)
)

print("\nFeature created successfully.")
print(
    enhanced_df["mean_same_period_last_year"].describe()
)
```

```python
# ============================================
# ENHANCED BAYESIAN LINEAR REGRESSION
# ============================================

enhanced_feature_cols = [
    "mean_7",
    "mean_30",
    "std_30",
    "mean_90",
    "std_90",
    "min_30",
    "max_30",
    "missing_ratio_30",
    "missing_ratio_90",
    "recent_change",
    "ratio_7_30",
    "ratio_30_90",
    "change_30_90",
    "relative_change_30_90",
    "cv_30",
    "max_to_mean_30",
    "mean_same_period_last_year"
]

X_enhanced = enhanced_df[enhanced_feature_cols].copy()
y_enhanced = np.log1p(enhanced_df["target_mean_30"])
flags_enhanced = enhanced_df["FLAG"]

print("Enhanced X shape:", X_enhanced.shape)

X_train_e, X_test_e, y_train_e, y_test_e, flag_train_e, flag_test_e = train_test_split(
    X_enhanced,
    y_enhanced,
    flags_enhanced,
    test_size=0.20,
    random_state=42,
    stratify=flags_enhanced
)

enhanced_bayesian_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", BayesianRidge())
])

enhanced_bayesian_model.fit(X_train_e, y_train_e)

print("Enhanced Bayesian model trained successfully!")
```

```python
# ============================================
# EVALUATE ENHANCED MODEL
# ============================================

enhanced_regressor = enhanced_bayesian_model.named_steps["model"]

X_test_e_imp = enhanced_bayesian_model.named_steps["imputer"].transform(X_test_e)
X_test_e_scaled = enhanced_bayesian_model.named_steps["scaler"].transform(X_test_e_imp)

y_pred_log_e, y_std_log_e = enhanced_regressor.predict(
    X_test_e_scaled,
    return_std=True
)

log_mae_e = mean_absolute_error(y_test_e, y_pred_log_e)
log_rmse_e = np.sqrt(mean_squared_error(y_test_e, y_pred_log_e))
log_r2_e = r2_score(y_test_e, y_pred_log_e)

print("Enhanced Bayesian Linear Regression")
print("------------------------------------")
print("MAE :", log_mae_e)
print("RMSE:", log_rmse_e)
print("R²  :", log_r2_e)
print("Mean uncertainty:", y_std_log_e.mean())
```

```python
# ============================================
# BAYESIAN ANOMALY SCORE
# ============================================

anomaly_df = pd.DataFrame(index=y_test_e.index)

# Actual and predicted values on log scale
anomaly_df["actual_log"] = y_test_e
anomaly_df["predicted_log"] = y_pred_log_e

# Bayesian predictive uncertainty
anomaly_df["prediction_std"] = y_std_log_e

# Signed residual
anomaly_df["residual"] = (
    anomaly_df["actual_log"]
    - anomaly_df["predicted_log"]
)

# Standardized residual
anomaly_df["z_score"] = (
    anomaly_df["residual"]
    / anomaly_df["prediction_std"]
)

# Absolute anomaly score
anomaly_df["anomaly_score"] = (
    anomaly_df["z_score"].abs()
)

# Ground-truth label for evaluation
anomaly_df["FLAG"] = flag_test_e

print("Anomaly dataframe shape:", anomaly_df.shape)

print("\nFirst 10 rows:")
print(anomaly_df.head(10))
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.base import clone

# ============================================
# OUT-OF-FOLD BAYESIAN ANOMALY FEATURES
# ============================================

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

oof_pred = np.zeros(len(X_train_e))
oof_std = np.zeros(len(X_train_e))

X_train_e_reset = X_train_e.reset_index(drop=True)
y_train_e_reset = y_train_e.reset_index(drop=True)
flag_train_e_reset = flag_train_e.reset_index(drop=True)

for fold, (train_idx, val_idx) in enumerate(
    skf.split(X_train_e_reset, flag_train_e_reset), 1
):
    fold_model = clone(enhanced_bayesian_model)

    fold_model.fit(
        X_train_e_reset.iloc[train_idx],
        y_train_e_reset.iloc[train_idx]
    )

    fold_regressor = fold_model.named_steps["model"]

    X_val_imp = fold_model.named_steps["imputer"].transform(
        X_train_e_reset.iloc[val_idx]
    )

    X_val_scaled = fold_model.named_steps["scaler"].transform(
        X_val_imp
    )

    pred, std = fold_regressor.predict(
        X_val_scaled,
        return_std=True
    )

    oof_pred[val_idx] = pred
    oof_std[val_idx] = std

    print(f"Fold {fold} completed")

oof_residual = y_train_e_reset - oof_pred

oof_z = oof_residual / oof_std

oof_anomaly = np.abs(oof_z)

train_anomaly_df = pd.DataFrame({
    "anomaly_score": oof_anomaly,
    "residual": oof_residual,
    "prediction_std": oof_std,
    "FLAG": flag_train_e_reset
})

print("\nOOF anomaly features created!")
print(train_anomaly_df.head())
```

```python
# ============================================
# UNIT 1 - PROBABILITY ANALYSIS
# ============================================

# Use the 95th percentile of NORMAL consumers
# as the high-anomaly threshold

normal_scores = anomaly_df.loc[
    anomaly_df["FLAG"] == 0,
    "anomaly_score"
]

anomaly_threshold = normal_scores.quantile(0.95)

print("95% normal-consumer anomaly threshold:",
      anomaly_threshold)

# Define high-anomaly event
high_anomaly = (
    anomaly_df["anomaly_score"] >= anomaly_threshold
)

# Prior probabilities
p_theft = (anomaly_df["FLAG"] == 1).mean()
p_normal = (anomaly_df["FLAG"] == 0).mean()

# Likelihoods
p_anomaly_given_theft = high_anomaly[
    anomaly_df["FLAG"] == 1
].mean()

p_anomaly_given_normal = high_anomaly[
    anomaly_df["FLAG"] == 0
].mean()

# Overall probability of high anomaly
p_anomaly = high_anomaly.mean()

# Bayes theorem
p_theft_given_anomaly = (
    p_anomaly_given_theft * p_theft
) / p_anomaly

print("\nProbability Analysis")
print("--------------------")
print("P(FLAG=1) =", p_theft)
print("P(FLAG=0) =", p_normal)

print("\nP(Anomaly | FLAG=1) =",
      p_anomaly_given_theft)

print("P(Anomaly | FLAG=0) =",
      p_anomaly_given_normal)

print("\nP(Anomaly) =",
      p_anomaly)

print("\nP(FLAG=1 | Anomaly) =",
      p_theft_given_anomaly)
```

```python
# ============================================
# UNIT 1 - POLYNOMIAL CURVE FITTING
# ============================================

poly_features = pd.DataFrame(index=model_df.index)

x = np.arange(len(recent_30_cols))

poly_rmse = []
poly_end_slope = []
poly_end_residual = []

for idx in model_df.index:

    values = clean_df.loc[idx, recent_30_cols].astype(float).values

    valid = ~np.isnan(values)

    # Need enough observations for a stable quadratic fit
    if valid.sum() < 10:
        poly_rmse.append(np.nan)
        poly_end_slope.append(np.nan)
        poly_end_residual.append(np.nan)
        continue

    x_valid = x[valid]
    y_valid = values[valid]

    # Degree-2 polynomial
    coefficients = np.polyfit(
        x_valid,
        y_valid,
        deg=2
    )

    polynomial = np.poly1d(coefficients)

    # Fitted values
    fitted = polynomial(x_valid)

    # RMSE of polynomial fit
    rmse = np.sqrt(
        np.mean((y_valid - fitted) ** 2)
    )

    # Derivative = trend slope
    derivative = np.polyder(polynomial)

    # Slope at end of 30-day period
    end_slope = derivative(x[-1])

    # Actual latest valid value
    last_x = x_valid[-1]
    last_y = y_valid[-1]

    end_residual = last_y - polynomial(last_x)

    poly_rmse.append(rmse)
    poly_end_slope.append(end_slope)
    poly_end_residual.append(end_residual)


poly_features["poly_rmse_30"] = poly_rmse
poly_features["poly_end_slope_30"] = poly_end_slope
poly_features["poly_end_residual_30"] = poly_end_residual

print("Polynomial features created!")

print(
    poly_features.describe()
)
```
